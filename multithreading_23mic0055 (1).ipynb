# using a threads in a sequential order

```python
#import concurrent.futures
import threading #using a threads in a sequential order 
import time

start=time.perf_counter()

def do_something():
    seconds=1
    print(f'sleeping {seconds} second(s)')
    time.sleep(1)
    print('done sleeping..')

t1=threading.Thread(target=do_something)

t1.start()#start
t1.join()
t2=threading.Thread(target=do_something)

t2.start()
t2.join()
finish=time.perf_counter()
print(f'Finished in{round(finish-start,2)} seconds(s)')
```

Here because of executing the threads sequentially the overall ime is higher comapartively with parallel processing this is the drawback of executing threads in sequentially 

# doing multithreading usinf for loop for parralel processing 

```python
import threading #doing multithreading usinf for loop for parralel processing 
import time

start=time.perf_counter()

def do_something():#function defintion 
    seconds=1
    print(f'sleeping {seconds} second(s)')
    time.sleep(1)#making sleep for 1 second
    print('done sleeping..')
threads=[]
for _ in range(10):
    t1=threading.Thread(target=do_something)
    t1.start()
    threads.append(t1)
for i in threads:
    i.join()
finish=time.perf_counter()
print(f'Finished in{round(finish-start,2)} seconds(s)')
```

using thread we acheived parallel processing because of that it took less time when compared with sequntial processing of threads 
and in sequential threading it took 2 seconds but in parallel it took 1 second because all the two threads started parallel excuted parallely so it took less time 

# doing multithreading usinf for loop for parralel processing 

```python
import threading #doing multithreading usinf for loop for parralel processing 
import time

start=time.perf_counter()

def do_something(seconds):#passing arguments 
    print(f'sleeping {seconds} second(s)')
    time.sleep(seconds)#passing the arguementvalue here 
    print('done sleeping..')
threads=[]
for _ in range(10):
    t1=threading.Thread(target=do_something,args=[2])
    t1.start()
    threads.append(t1)
for i in threads:
    i.join()
finish=time.perf_counter()
print(f'Finished in{round(finish-start,2)} seconds(s)')
```

in the last snippet we done parallel processing using for loop and here the same thing but there we used default as 1 second for each thread to get sleep but here i am passing the arguement and the thread gets slepped by the value of that passed arguement last snipped slept for 1 sec and in this snippet i am passing 2 as arguement so it slept for 2 seconds 

# doing multithreading using concurrent futures with the help of threadpool executor

```python
import concurrent.futures #doing multithreading using concurrent futures with the help of threadpool executor 

start=time.perf_counter()

def do_something(seconds):#function defintion 
   
    print(f'sleeping {seconds} second(s)')
    time.sleep(1)#making sleep for 1 second
    print('done sleeping..')
with concurrent.futures.ThreadPoolExecutor() as execuetor:
    f1=execuetor.submit(do_something,1.5)
    print(f1.result())
    
```

in this example we are using concurrent futures for parallel processing of threads because using concurrent futures it took less line of codes

 # doing multithreading using concurrent futures with the help of threadpool executor and functio with a single arguement  

```python
import concurrent.futures #doing multithreading using concurrent futures with the help of threadpool executor and functio with a single arguement  

start=time.perf_counter()

def do_something(seconds):#function defintion 
   
    print(f'sleeping {seconds} second(s)')
    time.sleep(seconds)#making sleep for some seconds which is mentioned in the arguements 
    return 'done sleeping..'
with concurrent.futures.ThreadPoolExecutor() as execuetor:
   
    res=[execuetor.submit(do_something,1.5) for _ in range(2)]#using list comprehension i am creating multiple threads 
    for f in concurrent.futures.as_completed(res):
       print(f.result())# this will start the excuetion 
finish=time.perf_counter()
print(f'Finished in{round(finish-start,2)} seconds(s)')
```

i am reducing the code by using list comprehension instead of for loop and also using oncureent futures of threadpoolexecutor,submit,result

# doing multithreading using concurrent futures with the help of threadpool executor and  with multiple arguements 

```python
import concurrent.futures #doing multithreading using concurrent futures with the help of threadpool executor and  with multiple arguements 

start=time.perf_counter()

def do_something(seconds):#function defintion 
   
    print(f'sleeping {seconds} second(s)')
    time.sleep(seconds)#making sleep for some seconds which is mentioned in the arguements 
    return f'done sleeping..{seconds}..'# prints th statemetn with the arguement of that particular thread to differentiate which thread gets finish off first 
with concurrent.futures.ThreadPoolExecutor() as execuetor:
    secs=[5,7,3,8,1]
    res=[execuetor.submit(do_something,sec) for sec in secs]#using list comprehension i am creating multiple threads 
    for f in concurrent.futures.as_completed(res):
       print(f.result())# this will start the excuetion 
finish=time.perf_counter()
print(f'Finished in{round(finish-start,2)} seconds(s)')
```

here i am passing a list of arguement during function call so each thread gets one by one value in the list for eac thread 

# doing multithreading using concurrent futures with the help of threadpool executor and map 

```python
import concurrent.futures #doing multithreading using concurrent futures with the help of threadpool executor and map 

start=time.perf_counter()

def do_something(seconds):#function defintion 
   
    print(f'sleeping {seconds} second(s)')
    time.sleep(seconds)#making sleep for some seconds which is mentioned in the arguements 
    return f'done sleeping..{seconds}..'# prints th statemetn with the arguement of that particular thread to differentiate which thread gets finish off first 
with concurrent.futures.ThreadPoolExecutor() as execuetor:
    secs=[1,7,3,4,5]
    res=execuetor.map(do_something,secs)#using map function i am creating multiple threads and executing it and printing stores the results in the res 
    for f in res:
       print(f)# printing the result that stored in res 
finish=time.perf_counter()
print(f'Finished in{round(finish-start,2)} seconds(s)')
```

i am using map function to map the arguement to each thread during function call beacuse of that it make the code simpler
